### Step 1: Import Libraries and Load Data

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the dataset from local storage
df = pd.read_csv('/Book1.csv')
display(df.head())
```

### Step 2: Exploratory Data Analysis

```python
sns.scatterplot(x=df['cgpa'], y=df['resume_score'], hue=df['placed'])
plt.show()
```

### Step 3: Feature-Target Splitting and Model Training

```python
from sklearn.linear_model import Perceptron

# Split features (X) and label (y)
X = df.iloc[:, 0:2]
y = df.iloc[:, -1]

# Instantiate and train the Perceptron
p = Perceptron()
p.fit(X, y)

# View coefficients
print("Model coefficients (Weights):", p.coef_)
print("Model intercept (Bias):", p.intercept_)
```

### Step 4: Decision Boundary Visualization

```python
from mlxtend.plotting import plot_decision_regions
import matplotlib.pyplot as plt

# Plot the decision regions using mlxtend (passing values to avoid feature name warnings)
plot_decision_regions(X.values, y.values, clf=p, legend=2)

# Label the axes
plt.xlabel('CGPA')
plt.ylabel('Resume Score')
plt.title('Perceptron Decision Boundary')
plt.show()
```
